```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# ==============================================================================
# STEP 3: SPICE Output Parser, Data Extractor & Transient Waveform Plotting
# ==============================================================================
def parse_ngspice_tran(filename):
    time_list, v_out_list = [], []
    num_regex = r"[-+]?\d*\.\d+e[-+]?\d+|[-+]?\d+\.\d*"
    if not os.path.exists(filename):
        return None, None
    with open(filename, "r") as f:
        for line in f:
            tokens = line.split()
            if len(tokens) >= 3:
                if re.match(num_regex, tokens[1]) and re.match(num_regex, tokens[2]):
                    try:
                        time_list.append(float(tokens[1]))
                        v_out_list.append(float(tokens[2]))
                    except ValueError:
                        continue
    return np.array(time_list), np.array(v_out_list)

t_correct, v_correct = parse_ngspice_tran("out_tran_1111.txt")
t_wrong, v_wrong = parse_ngspice_tran("out_tran_0000.txt")

if t_correct is not None and t_wrong is not None:
    df_correct = pd.DataFrame({'Time': t_correct, 'V_out': v_correct})
    df_wrong = pd.DataFrame({'Time': t_wrong, 'V_out': v_wrong})
    df_correct.to_csv("dataset_transient.csv", index=False)
    df_wrong.to_csv("dataset_transient_wrong.csv", index=False)

    fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(10, 6.5), sharex=True)

    ax1.plot(t_correct * 1e9, v_correct, color='darkblue', linewidth=1.5, label='Correct Key [1, 1, 1, 1]')
    ax1.set_title('Hardware Obfuscation Evaluation: Nominal / Active State (Correct Key)', fontsize=11, fontweight='bold')
    ax1.set_ylabel('Output Voltage (V)', fontsize=10)
    ax1.grid(True, linestyle='--', alpha=0.6)
    ax1.legend(loc='lower right')

    ax2.plot(t_wrong * 1e9, v_wrong, color='crimson', linewidth=1.5, label='Locked Key [0, 0, 0, 0]')
    ax2.set_title('Hardware Obfuscation Evaluation: Obfuscated / Locked State (Wrong Key)', fontsize=11, fontweight='bold')
    ax2.set_xlabel('Time (ns)', fontsize=10)
    ax2.set_ylabel('Output Voltage (V)', fontsize=10)
    ax2.grid(True, linestyle='--', alpha=0.6)
    ax2.legend(loc='lower right')

    plt.tight_layout()
    plt.savefig("transient_comparison_conference.png", dpi=300)
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.ensemble import RandomForestClassifier
from sklearn.svm import SVC
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score

# ==============================================================================
# STEP 4: Scaled Feature Extraction & ML Attack Robustness Analysis
# ==============================================================================
df_correct = pd.read_csv('dataset_transient.csv')
df_wrong = pd.read_csv('dataset_transient_wrong.csv')

def create_windowed_dataset(df_active, df_locked, window_size=20):
    features, labels = [], []
    v_act = df_active['V_out'].values
    for i in range(0, len(v_act) - window_size, 5):
        win = v_act[i : i + window_size]
        features.append([np.mean(win), np.std(win), np.ptp(win)])
        labels.append(1)

    v_lock = df_locked['V_out'].values
    for i in range(0, len(v_lock) - window_size, 5):
        win = v_lock[i : i + window_size]
        features.append([np.mean(win), np.std(win), np.ptp(win)])
        labels.append(0)

    return np.array(features), np.array(labels)

X_clean, y = create_windowed_dataset(df_correct, df_wrong, window_size=25)
noise_levels = np.linspace(0.0, 0.15, 20)

rf_acc_history, svm_acc_history, mlp_acc_history = [], [], []

print("🔄 Executing Scaled ML Attack Robustness Analysis...")
for noise_std in noise_levels:
    np.random.seed(42)
    noise = np.random.normal(0, noise_std, X_clean.shape)
    X_noisy = X_clean + noise

    X_train, X_test, y_train, y_test = train_test_split(X_noisy, y, test_size=0.25, random_state=42)
    scaler = StandardScaler()
    X_train_scaled = scaler.fit_transform(X_train)
    X_test_scaled = scaler.transform(X_test)

    clf_rf = RandomForestClassifier(n_estimators=50, random_state=42)
    clf_rf.fit(X_train_scaled, y_train)
    rf_acc_history.append(accuracy_score(y_test, clf_rf.predict(X_test_scaled)))

    clf_svm = SVC(kernel='rbf', C=10.0, random_state=42)
    clf_svm.fit(X_train_scaled, y_train)
    svm_acc_history.append(accuracy_score(y_test, clf_svm.predict(X_test_scaled)))

    clf_mlp = MLPClassifier(hidden_layer_sizes=(20, 10), max_iter=700, random_state=42)
    clf_mlp.fit(X_train_scaled, y_train)
    mlp_acc_history.append(accuracy_score(y_test, clf_mlp.predict(X_test_scaled)))

plt.figure(figsize=(10, 5.5))
plt.plot(noise_levels * 1000, [a * 100 for a in rf_acc_history], color='darkviolet', marker='o', label='Random Forest (Robust Baseline)')
plt.plot(noise_levels * 1000, [a * 100 for a in svm_acc_history], color='darkorange', marker='s', label='Support Vector Machine (SVM)')
plt.plot(noise_levels * 1000, [a * 100 for a in mlp_acc_history], color='forestgreen', marker='^', label='Neural Network (MLP)')
plt.axhline(y=50, color='gray', linestyle=':', label='Random Guessing (50%)')
plt.title('ML Side-Channel Attack Success Rate vs. Measurement Noise', fontsize=12, fontweight='bold')
plt.xlabel('Noise Standard Deviation (mV)', fontsize=10)
plt.ylabel('Attack Classification Accuracy (%)', fontsize=10)
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend(loc='lower left')
plt.tight_layout()
plt.savefig("ml_benchmarking_final_scaled.png", dpi=300)
plt.show()
```

```python
# ==============================================================================
# STEP 6: Monte Carlo SPICE Mismatch & Yield Analysis (65nm TSMC - 100 Runs)
# ==============================================================================
import os

def run_monte_carlo_batch(n_runs=100, target_key=[1, 1, 1, 1]):
    gains = []
    v_k = [1.2 if b == 1 else 0.0 for b in target_key]

    for r in range(n_runs):
        np.random.seed(r * 7 + 13)
        vto_m = np.random.normal(0.35, 0.35 * 0.04)   # +-4% Vth variation
        kp_m = np.random.normal(120e-6, 120e-6 * 0.03) # +-3% Kp variation

        mc_spice = f"""* Monte Carlo Run {r} (TSMC 65nm)
.model nmos_mc nmos (level=1 vto={vto_m} kp={kp_m} cgso=2n cgdo=2n)
.param LO_FREQ=2.4G RF_FREQ=2.3G VDD_VAL=1.2
.param V_K0={v_k[0]} V_K1={v_k[1]} V_K2={v_k[2]} V_K3={v_k[3]}

Vdc net2 0 DC {{VDD_VAL}}
V3 net20 0 DC 0.600 sin(0.600 200m {{LO_FREQ}})
V2 net8  0 DC 0.600 sin(0.600 200m {{LO_FREQ}})
V1 net7  0 DC 0.800 sin(0.800 50m {{RF_FREQ}})
V0 net9  0 DC 0.800 sin(0.800 50m {{RF_FREQ}})

RL2 net2 net6 200
RL1 net2 net10 200

M5 net10 net7 net11 0 nmos_mc w=5u l=65n
M6 net6 net9 net11 0 nmos_mc w=5u l=65n
M3 net10 net9 net3 0 nmos_mc w=5u l=65n
M4 net6 net7 net3 0 nmos_mc w=5u l=65n
M2 net11 net20 net19 0 nmos_mc w=5u l=65n
M1 net3 net8 net19 0 nmos_mc w=5u l=65n

Mt0 net19 V_K0_node 0 0 nmos_mc w=1u l=65n
Mt1 net19 V_K1_node 0 0 nmos_mc w=2u l=65n
Mt2 net19 V_K2_node 0 0 nmos_mc w=4u l=65n
Mt3 net19 V_K3_node 0 0 nmos_mc w=8u l=65n

Vk0 V_K0_node 0 DC {{V_K0}}
Vk1 V_K1_node 0 DC {{V_K1}}
Vk2 V_K2_node 0 DC {{V_K2}}
Vk3 V_K3_node 0 DC {{V_K3}}

.tran 0.01n 6n
.print tran v(net6)
.end
"""
        fname = f"mc_temp_{r}.sp"
        out_name = f"mc_out_{r}.txt"
        with open(fname, "w") as f:
            f.write(mc_spice)
        os.system(f"ngspice -b {fname} > {out_name}")

        _, v_mc = parse_ngspice_tran(out_name)
        if v_mc is not None:
            v_pp = np.ptp(v_mc[-300:])
            gain_val = 20 * np.log10(max((v_pp / 2.0), 1e-6) / 0.05)
            gains.append(gain_val)

    return np.array(gains)

print("🔄 Running 100 Monte Carlo variations for Nominal Active Key [1,1,1,1]...")
mc_corr = run_monte_carlo_batch(n_runs=100, target_key=[1, 1, 1, 1])

print("🔄 Running 100 Monte Carlo variations for Locked Key [0,0,0,0]...")
mc_wrong = run_monte_carlo_batch(n_runs=100, target_key=[0, 0, 0, 0])

# --- Σχεδίαση Ιστογράμματος ---
plt.figure(figsize=(8.5, 4.5), dpi=120)
plt.hist(mc_wrong, bins=20, alpha=0.75, color='#d62728', edgecolor='black', label='Locked Key [0,0,0,0]')
plt.hist(mc_corr, bins=20, alpha=0.75, color='#1f77b4', edgecolor='black', label='Nominal Active Key [1,1,1,1]')

plt.title('Monte Carlo Process & Mismatch Variation (100 SPICE Runs)', fontsize=11, fontweight='bold')
plt.xlabel('Differential Conversion Gain (dB)', fontsize=10, fontweight='bold')
plt.ylabel('Instance Count', fontsize=10, fontweight='bold')
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(loc='upper center', framealpha=0.9)

plt.tight_layout()
plt.savefig('monte_carlo_yield_real_100.png', dpi=300)
plt.show()
print("✅ Monte Carlo 100-run analysis complete! Plot saved as 'monte_carlo_yield_real_100.png'.")
```

```python
# ==============================================================================
# STEP 8: PVT (Process, Voltage, Temperature) Corner Stability Analysis
# ==============================================================================
pvt_corners = [
    {"name": "TT / 27°C / 1.20V",  "temp": 27,  "vdd": 1.20, "vto": 0.35, "kp": 120e-6},
    {"name": "SS / 125°C / 1.08V", "temp": 125, "vdd": 1.08, "vto": 0.39, "kp": 105e-6},
    {"name": "FF / -40°C / 1.32V", "temp": -40, "vdd": 1.32, "vto": 0.31, "kp": 135e-6},
    {"name": "FS / 85°C / 1.08V",  "temp": 85,  "vdd": 1.08, "vto": 0.31, "kp": 105e-6},
    {"name": "SF / 85°C / 1.32V",  "temp": 85,  "vdd": 1.32, "vto": 0.39, "kp": 135e-6}
]

def run_pvt_sim(target_key=[1, 1, 1, 1]):
    gains = []
    v_k = [1.2 if b == 1 else 0.0 for b in target_key]

    for corner in pvt_corners:
        spice_pvt = f"""* PVT Corner Analysis: {corner['name']} (TSMC 65nm)
.model nmos_pvt nmos (level=1 vto={corner['vto']} kp={corner['kp']} cgso=2n cgdo=2n)
.temp {corner['temp']}
.param LO_FREQ=2.4G RF_FREQ=2.3G VDD_VAL={corner['vdd']}
.param V_K0={v_k[0]} V_K1={v_k[1]} V_K2={v_k[2]} V_K3={v_k[3]}

Vdc net2 0 DC {{VDD_VAL}}
V3 net20 0 DC 0.600 sin(0.600 200m {{LO_FREQ}})
V2 net8  0 DC 0.600 sin(0.600 200m {{LO_FREQ}})
V1 net7  0 DC 0.800 sin(0.800 50m {{RF_FREQ}})
V0 net9  0 DC 0.800 sin(0.800 50m {{RF_FREQ}})

RL2 net2 net6 200
RL1 net2 net10 200

M5 net10 net7 net11 0 nmos_pvt w=5u l=65n
M6 net6 net9 net11 0 nmos_pvt w=5u l=65n
M3 net10 net9 net3 0 nmos_pvt w=5u l=65n
M4 net6 net7 net3 0 nmos_pvt w=5u l=65n
M2 net11 net20 net19 0 nmos_pvt w=5u l=65n
M1 net3 net8 net19 0 nmos_pvt w=5u l=65n

Mt0 net19 V_K0_node 0 0 nmos_pvt w=1u l=65n
Mt1 net19 V_K1_node 0 0 nmos_pvt w=2u l=65n
Mt2 net19 V_K2_node 0 0 nmos_pvt w=4u l=65n
Mt3 net19 V_K3_node 0 0 nmos_pvt w=8u l=65n

Vk0 V_K0_node 0 DC {{V_K0}}
Vk1 V_K1_node 0 DC {{V_K1}}
Vk2 V_K2_node 0 DC {{V_K2}}
Vk3 V_K3_node 0 DC {{V_K3}}

.tran 0.01n 6n
.print tran v(net6)
.end
"""
        fname = "pvt_temp.sp"
        out_name = "pvt_out.txt"
        with open(fname, "w") as f:
            f.write(spice_pvt)
        os.system(f"ngspice -b {fname} > {out_name}")

        _, v_pvt = parse_ngspice_tran(out_name)
        if v_pvt is not None:
            v_pp = np.ptp(v_pvt[-300:])
            gain = 20 * np.log10(max((v_pp / 2.0), 1e-6) / 0.05)
            gains.append(gain)
        else:
            gains.append(-25.0)

    return gains

print("🔄 Running PVT Analysis for Nominal Active Key [1,1,1,1]...")
pvt_corr = run_pvt_sim(target_key=[1, 1, 1, 1])
print("🔄 Running PVT Analysis for Locked Key [0,0,0,0]...")
pvt_wrong = run_pvt_sim(target_key=[0, 0, 0, 0])

corner_labels = [c["name"] for c in pvt_corners]
x_pos = np.arange(len(corner_labels))
width = 0.35

fig, ax = plt.subplots(figsize=(9.5, 4.8), dpi=120)
ax.bar(x_pos - width/2, pvt_corr, width, label='Nominal Active Key [1,1,1,1]', color='#1f77b4', edgecolor='black')
ax.bar(x_pos + width/2, pvt_wrong, width, label='Locked Key [0,0,0,0]', color='#d62728', edgecolor='black')

ax.set_ylabel('Differential Conversion Gain (dB)', fontsize=10, fontweight='bold')
ax.set_title('PVT Corner Stability & Security Gap Verification', fontsize=11, fontweight='bold')
ax.set_xticks(x_pos)
ax.set_xticklabels(corner_labels, rotation=15, ha='right', fontsize=9)
ax.axhline(0, color='black', linewidth=0.8, linestyle='--')
ax.grid(True, linestyle=':', alpha=0.6)
ax.legend(loc='lower left', framealpha=0.9)

plt.tight_layout()
plt.savefig('pvt_corners_comparison.png', dpi=300)
plt.show()
print("✅ Step 8 complete! Plot saved as 'pvt_corners_comparison.png'.")
```
